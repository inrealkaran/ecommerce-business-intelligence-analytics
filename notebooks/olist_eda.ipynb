```python
import pandas as pd
import numpy as np
import os
import zipfile
```

```python
zip_file = list(uploaded.keys())[0]

extract_path = "/content/olist_data"

with zipfile.ZipFile(zip_file, "r") as zip_ref:
    zip_ref.extractall(extract_path)

print("Dataset extracted successfully!")
```

```python
csv_files = []

for root, dirs, files_list in os.walk(extract_path):
    for file in files_list:
        if file.endswith(".csv"):
            csv_files.append(os.path.join(root, file))

tables = {}

for file in csv_files:
    table_name = os.path.basename(file).replace(".csv", "")
    tables[table_name] = pd.read_csv(file)

print("Tables loaded:", len(tables))
print("\nTable names:")
print(list(tables.keys()))
```

```python
missing_summary = []

for name, df in tables.items():
    total_rows = len(df)

    for col in df.columns:
        missing_count = df[col].isna().sum()

        if missing_count > 0:
            missing_summary.append({
                "Table": name,
                "Column": col,
                "Missing_Count": missing_count,
                "Missing_Percentage": round(
                    (missing_count / total_rows) * 100, 2
                )
            })

missing_df = pd.DataFrame(missing_summary)

if len(missing_df) > 0:
    missing_df = missing_df.sort_values(
        by="Missing_Percentage",
        ascending=False
    )
    display(missing_df)
else:
    print("No missing values found.")
```

```python
print("========== TABLE SUMMARY ==========")

for name, df in tables.items():
    print(f"{name}: {df.shape[0]} rows, {df.shape[1]} columns")

print("\n========== MISSING VALUES ==========")

for name, df in tables.items():
    missing = df.isnull().sum()
    missing = missing[missing > 0]

    if len(missing) > 0:
        print(f"\n{name}")
        print(missing.sort_values(ascending=False))

print("\n========== DUPLICATES ==========")

for name, df in tables.items():
    print(f"{name}: {df.duplicated().sum()} duplicate rows")

print("\n========== DATE RANGES ==========")

for name, df in tables.items():
    for col in df.columns:
        if "date" in col.lower() or "time" in col.lower():
            dates = pd.to_datetime(df[col], errors="coerce")
            print(f"{name} | {col} | {dates.min()} to {dates.max()}")
```

```python
geo = tables["olist_geolocation_dataset"]

print("Total rows:", len(geo))

print("Exact duplicate rows:", geo.duplicated().sum())

print("Unique postal codes:",
      geo["geolocation_zip_code_prefix"].nunique())

print("Unique coordinate pairs:",
      geo[["geolocation_lat", "geolocation_lng"]]
      .drop_duplicates().shape[0])

print("\nDuplicate postal codes:")
print(geo["geolocation_zip_code_prefix"].duplicated().sum())

print("\nDuplicate coordinate pairs:")
print(geo[["geolocation_lat", "geolocation_lng"]]
    .duplicated().sum())
```

```python
geo = tables["olist_geolocation_dataset"].copy()

geo["coordinate_pair"] = list(zip(
    geo["geolocation_lat"],
    geo["geolocation_lng"]
))

zip_analysis = geo.groupby("geolocation_zip_code_prefix").agg(
    total_records=("geolocation_zip_code_prefix", "size"),
    unique_coordinates=("coordinate_pair", "nunique"),
    unique_cities=("geolocation_city", "nunique"),
    unique_states=("geolocation_state", "nunique")
).reset_index()

print("Postal codes:", len(zip_analysis))

print("\nPostal codes with multiple coordinate pairs:")
display(
    zip_analysis[
        zip_analysis["unique_coordinates"] > 1
    ].sort_values("total_records", ascending=False).head(15)
)

print("\nPostal codes mapped to multiple cities:")
display(
    zip_analysis[
        zip_analysis["unique_cities"] > 1
    ].sort_values("unique_cities", ascending=False).head(15)
)

print("\nPostal codes mapped to multiple states:")
display(
    zip_analysis[
        zip_analysis["unique_states"] > 1
    ].sort_values("unique_states", ascending=False).head(15)
)
```

```python
items = tables["olist_order_items_dataset"].copy()
orders = tables["olist_orders_dataset"].copy()

items["shipping_limit_date"] = pd.to_datetime(items["shipping_limit_date"])
orders["order_purchase_timestamp"] = pd.to_datetime(
    orders["order_purchase_timestamp"]
)

late_items = items[
    items["shipping_limit_date"] >= "2019-01-01"
]

result = late_items.merge(
    orders[
        [
            "order_id",
            "order_status",
            "order_purchase_timestamp",
            "order_approved_at",
            "order_delivered_carrier_date",
            "order_delivered_customer_date"
        ]
    ],
    on="order_id",
    how="left"
)

display(result.sort_values("shipping_limit_date"))
```

```python
orders = tables["olist_orders_dataset"]
items = tables["olist_order_items_dataset"]

orders["order_purchase_timestamp"] = pd.to_datetime(orders["order_purchase_timestamp"])

sales = items.merge(
    orders[["order_id", "order_purchase_timestamp"]],
    on="order_id"
)

sales["month"] = sales["order_purchase_timestamp"].dt.to_period("M")

monthly_sales = sales.groupby("month")["price"].sum().reset_index()

monthly_sales.columns = ["Month", "Revenue"]

display(monthly_sales)
```

```python
sept = sales[sales["order_purchase_timestamp"].dt.to_period("M") == "2018-09"]

print("September orders:", sept["order_id"].nunique())
print("September items:", len(sept))
print("September revenue:", sept["price"].sum())
```

```python
items = tables["olist_order_items_dataset"]
products = tables["olist_products_dataset"]
translation = tables["product_category_name_translation"]

category_sales = items.merge(products, on="product_id")
category_sales = category_sales.merge(
    translation, on="product_category_name", how="left"
)

category_revenue = (
    category_sales.groupby("product_category_name_english")["price"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
    .reset_index()
)

category_revenue.columns = ["Category", "Revenue"]

display(category_revenue)
```

```python
category_orders = (
    category_sales.groupby("product_category_name_english")["order_id"]
    .nunique()
    .sort_values(ascending=False)
    .head(10)
    .reset_index()
)

category_orders.columns = ["Category", "Orders"]

display(category_orders)
```

```python
orders = tables["olist_orders_dataset"].copy()

orders["order_delivered_customer_date"] = pd.to_datetime(
    orders["order_delivered_customer_date"]
)

orders["order_estimated_delivery_date"] = pd.to_datetime(
    orders["order_estimated_delivery_date"]
)

delivered = orders[orders["order_status"] == "delivered"].copy()

delivered["delivery_delay_days"] = (
    delivered["order_delivered_customer_date"] -
    delivered["order_estimated_delivery_date"]
).dt.days

print("Delivered orders:", len(delivered))
print("Late orders:", (delivered["delivery_delay_days"] > 0).sum())
print("On-time/early orders:", (delivered["delivery_delay_days"] <= 0).sum())
print("Average delay (days):", delivered["delivery_delay_days"].mean())
```

```python
reviews = tables["olist_order_reviews_dataset"]

review_analysis = delivered.merge(
    reviews[["order_id", "review_score"]],
    on="order_id",
    how="inner"
)

review_analysis["delivery_status"] = review_analysis["delivery_delay_days"].apply(
    lambda x: "Late" if x > 0 else "On time/Early"
)

result = review_analysis.groupby("delivery_status").agg(
    Orders=("order_id", "nunique"),
    Average_Review=("review_score", "mean")
).reset_index()

display(result)
```

```python
late_category = review_analysis.merge(
    items[["order_id", "product_id"]].drop_duplicates(),
    on="order_id",
    how="left"
)

late_category = late_category.merge(
    products[["product_id", "product_category_name"]],
    on="product_id",
    how="left"
)

late_category = late_category.drop_duplicates(
    ["order_id", "product_category_name"]
)

late_category["is_late"] = late_category["delivery_delay_days"] > 0

category_delay = late_category.groupby("product_category_name").agg(
    Orders=("order_id", "nunique"),
    Late_Orders=("is_late", "sum")
)

category_delay["Late_Rate_%"] = (
    category_delay["Late_Orders"] / category_delay["Orders"] * 100
)

display(category_delay.sort_values("Late_Rate_%", ascending=False).head(10))
```

```python
payments = tables["olist_order_payments_dataset"]

payment_analysis = payments.groupby("payment_type").agg(
    Transactions=("order_id", "count"),
    Total_Value=("payment_value", "sum")
).sort_values("Transactions", ascending=False)

display(payment_analysis)
```

```python
customers = tables["olist_customers_dataset"]

customer_data = customer_orders.merge(
    customers[["customer_id", "customer_unique_id"]],
    on="customer_id",
    how="left"
)

customer_frequency = customer_data.groupby("customer_unique_id")["order_id"].nunique()

repeat_customers = (customer_frequency > 1).sum()
total_customers = customer_frequency.count()

print("Unique Customers:", total_customers)
print("Repeat Customers:", repeat_customers)
print("Repeat Customer Rate:", round(repeat_customers / total_customers * 100, 2), "%")
```
